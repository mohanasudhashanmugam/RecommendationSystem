```python
#1.Answers:
import pandas as pd
from sklearn.feature_selection import RFE
from sklearn.linear_model import LogisticRegression

X = [[1, 2], [3, 4], [5, 6]]
y = [1, 0, 1]

#  Convert X into a DataFrame with custom column names
X_df = pd.DataFrame(X, columns=['feature_1', 'feature_2'])

y_df = pd.Series(y, name='target')


rfe = RFE(estimator=LogisticRegression(), n_features_to_select=1)
X_new = rfe.fit_transform(X_df, y_df)

selected_features = X_df.columns[rfe.support_].tolist()  

print("Selected features:",selected_features)


```

```python
#4.Answers

from sklearn.feature_selection import SelectFromModel
from sklearn.tree import DecisionTreeClassifier

X = [[1, 2], [3, 4], [5, 6]]
y = [1, 0, 1]

x=pd.DataFrame(X, columns=['feature1','feature2'])

selector = SelectFromModel(DecisionTreeClassifier(), threshold="mean")
X_new = selector.fit_transform(X, y)

selected_features = selector.get_feature_names_out(input_features=x.columns)
print("Selected features:" ,list(selected_features) )
```
